### Source T03: Fourth-quarter 2026 tariffs

Organisation: Public Utilities Regulatory Commission (PURC)

Relevant section:
Fourth Schedule — End-User Tariffs, PDF page 3.

The schedule distinguishes Non-Residential and Special Load
Tariff (SLT) customer categories. Electricity rates are quoted
in Ghana pesewas, so conversion to Ghana cedis requires
division by 100.

Next verification:
Select a customer category and document its energy and service
charges. I do not assume the model's 100 kWp solar size determines
the customer's electricity tariff category.

```python
import pandas as pd

business_tariffs = pd.DataFrame({
    "Customer category": [
        "Non-Residential: 0–300",
        "Non-Residential: 301+",
        "SLT-LV"
    ],
    "Energy charge (GHp/kWh)": [
        183.9549,
        224.0395,
        240.2104
    ],
    "Service charge (GHp/month)": [
        1242.8245,
        1242.8245,
        50000.0000
    ]
})

business_tariffs["Energy charge (GHS/kWh)"] = (
    business_tariffs["Energy charge (GHp/kWh)"] / 100
)

business_tariffs["Service charge (GHS/month)"] = (
    business_tariffs["Service charge (GHp/month)"] / 100
)

business_tariffs["Source ID"] = "T03"
business_tariffs["Effective date"] = "2026-10-01"

business_tariffs.round(4)
```

### Tariff selection and solar-savings treatment

The table records selected published tariff rates, rather than
a confirmed tariff for a particular customer.

For an illustrative case, we will examine an SLT-LV customer.
This customer classification is an assumption until confirmed
by an electricity bill or utility documentation.

The initial savings calculation will use the published energy
charge for electricity purchases displaced by on-site solar.
The monthly service charge will remain payable in the model
and will not be counted as a solar saving.

Taxes, levies and any other bill adjustments require separate
verification before estimating total bill savings.

The tariff is denominated in Ghana cedis. We must document an
exchange rate and its date before comparing it with notebook 03's
USD assumptions.

```python
# Illustrative system assumptions carried over from notebook 03
system_size_kwp = 100
annual_yield_kwh_per_kwp = 1500
self_consumption_rate = 0.90

# Retrieve the SLT-LV energy rate from the documented tariff table
slt_lv_energy_rate = business_tariffs.loc[
    business_tariffs["Customer category"] == "SLT-LV",
    "Energy charge (GHS/kWh)"
].iloc[0]

annual_generation_kwh = (
    system_size_kwp * annual_yield_kwh_per_kwp
)
annual_self_consumed_kwh = (
    annual_generation_kwh * self_consumption_rate
)
annual_energy_charge_savings_ghs = (
    annual_self_consumed_kwh * slt_lv_energy_rate
)

print(f"Energy rate: GHS {slt_lv_energy_rate:.6f}/kWh")
print(f"Annual generation: {annual_generation_kwh:,.0f} kWh")
print(f"Annual on-site use: {annual_self_consumed_kwh:,.0f} kWh")
print(
    "Annual energy-charge savings: "
    f"GHS {annual_energy_charge_savings_ghs:,.2f}"
)
```

### Illustrative SLT-LV energy-charge savings

Using the documented SLT-LV energy charge of GHS 2.402104/kWh,
135,000 kWh of annual on-site solar consumption produces estimated
annual energy-charge savings of GHS 324,284.04.

The tariff is sourced from T03, effective 1 October 2026.
System size, solar yield, self-consumption and customer
classification remain illustrative assumptions.

This annual estimate holds the October 2026 rate constant for
12 months. It is not an estimate of actual calendar-year 2026
savings or a forecast of future tariffs.

The calculation excludes maintenance, financing, taxes, levies
and other bill adjustments. The service charge remains payable
in the model. Therefore, this figure represents gross
energy-charge savings rather than net project savings.

### Currency conversion method

The Ghana tariff and energy-charge savings are expressed in GHS.
Notebook 03 expresses installation costs and savings in USD.

For a dated comparison, we will obtain a Bank of Ghana reference
exchange rate and record its date, source and rate type.

If the exchange rate is quoted as GHS per USD:

USD equivalent = GHS amount / exchange rate

This conversion provides a snapshot comparison. It does not
forecast exchange rates over the project's lifetime.

The original illustrative USD model will remain separate from
the Ghana tariff case so their assumptions can be compared clearly.

### Source FX01: USD/GHS reference exchange rate

Organisation: Bank of Ghana
Page: Daily Interbank FX Rates
URL:
https://www.bog.gov.gh/treasury-and-the-markets/daily-interbank-fx-rates/

Rate date: 16 October 2026
Accessed: 18 October 2026
Currency pair: USDGHS
Buying rate: 11.6592 GHS/USD
Selling rate: 11.6708 GHS/USD
Mid rate: 11.6650 GHS/USD

The figures were transcribed from the displayed source row.
The mid rate will be used for a dated comparison:

USD equivalent = GHS amount / 11.6650

This reference conversion excludes transaction fees and does
not forecast future exchange rates.

```python
# FX01: Bank of Ghana mid rate, 16 October 2026
exchange_rate_ghs_per_usd = 11.6650

slt_lv_energy_rate_usd = (
    slt_lv_energy_rate / exchange_rate_ghs_per_usd
)

annual_energy_charge_savings_usd = (
    annual_energy_charge_savings_ghs / exchange_rate_ghs_per_usd
)

print(f"Exchange rate: {exchange_rate_ghs_per_usd:.4f} GHS/USD")
print(f"SLT-LV energy charge: USD {slt_lv_energy_rate_usd:.4f}/kWh")
print(
    "Gross annual energy-charge savings: "
    f"USD {annual_energy_charge_savings_usd:,.2f}"
)
```

```python
original_electricity_value_usd = 0.15

tariff_comparison = pd.DataFrame({
    "Case": [
        "Notebook 03 illustrative assumption",
        "SLT-LV tariff converted using FX01"
    ],
    "Electricity value (USD/kWh)": [
        original_electricity_value_usd,
        slt_lv_energy_rate_usd
    ],
    "Gross annual energy savings (USD)": [
        annual_self_consumed_kwh * original_electricity_value_usd,
        annual_energy_charge_savings_usd
    ]
})

tariff_comparison.round(4)
```

### Comparison with the original model

Notebook 03 assumes an avoided electricity cost of USD 0.15/kWh.
The documented SLT-LV tariff, converted using the exchange rate
dated 16 October 2026, is approximately USD 0.2059/kWh.

With annual on-site solar consumption held at 135,000 kWh,
the converted tariff produces higher gross energy-charge savings
than the original assumption.

This is a snapshot comparison using a published tariff and a
dated exchange rate. Customer classification, solar generation
and self-consumption remain assumptions. Maintenance and other
bill adjustments are excluded.

The comparison does not establish actual project profitability
or predict future USD savings.

```python
# Convert the USD savings back to GHS
savings_converted_back_ghs = (
    annual_energy_charge_savings_usd * exchange_rate_ghs_per_usd
)

conversion_difference = (
    savings_converted_back_ghs - annual_energy_charge_savings_ghs
)

print(
    f"Original GHS savings: "
    f"{annual_energy_charge_savings_ghs:,.2f}"
)
print(
    f"Converted back to GHS: "
    f"{savings_converted_back_ghs:,.2f}"
)
print(f"Conversion difference: GHS {conversion_difference:.8f}")
```
