## Quotation comparability checklist

Before calculating cost per kWp, confirm:

1. Capacity
   The quoted capacity refers to panel DC capacity in kWp.
   Record inverter AC capacity separately.

2. System type
   Identify whether batteries are included. Do not compare a
   solar-plus-storage package directly with a PV-only assumption.

3. Installation scope
   Check mounting, electrical works, protection, transport,
   commissioning and any roof reinforcement.

4. Additional costs
   Record taxes, metering, connection fees and exclusions.
   Unknown costs must remain marked as unknown.

5. Price basis
   Record currency, quotation date, validity and payment terms.
   Any currency conversion needs a documented, dated rate.

6. Operating costs
   Record maintenance, warranties and expected replacements.

Decision:
Classify each quotation as comparable, comparable after documented
adjustments, or insufficiently specified.

A single quotation is project evidence, not a Ghana market average.
